```python
X_train,X_test,y_teain,y_test=train_test_split(
    data.drop(labels=['target'],axis=1),
    data['target'],
    test_size=.3,
    random_state=42
)
print(f"The shape of traninig data is : {X_train.shape}")
print(f"\nThe shape of testing data is : {X_test.shape}")
```

# Applying Variance Threshold (scikit-learn)

In this step, we apply the **Variance Threshold** technique from the `sklearn.feature_selection` module to detect and remove **zero-variance features**.


After fitting the selector, the `get_support()` method is used to identify **which features are selected and which are removed** based on the variance threshold.

```python
selecte=VarianceThreshold(threshold=0)
selecte.fit(X_train) 
```

```python
print(sum(selecte.get_support()))
```

```python
# numer of removed_features 
costant_features=X_train.columns[~selecte.get_support()]
len(costant_features)
```

```python
print("the Zero_variance columns are : \n")
costant_features
```

```python
# to prof our cliam that zero variance is the constant feature
for col in costant_features:
    print(col,X_train[col].unique())
```

### Removing the zero variance features on `X_train` and `X_test`

```python
# get columns fame 
features_name=X_train.columns[selecte.get_support()]
```

```python
X_train=selecte.transform(X_train)
X_test=selecte.transform(X_test)
print(f"the shape of X_train is {X_train.shape} \nand the X_test shape is {X_test.shape}")
```
